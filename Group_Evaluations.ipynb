```python
## General imports
import os
import pandas as pd
```

```python
## Directories
dir_evals = './all_params/eval_model_by_model/'
dir_MVLR  = './all_params/all_MVLR_fitting_results.csv'
dir_MVPR  = './all_params/all_MVPR_fitting_results.csv'
dir_FNO   = './all_params/all_FNO_fitting_results.csv'

## Variables
list_MVLR = []
list_MVPR = []
list_FNO  = []
```

```python
## Get data
for fileName in os.listdir(dir_evals):
    pathFile = dir_evals + fileName
    data_file = pd.read_csv(pathFile, keep_default_na = False, na_values = ['NaN', 'N/A', ''])
    if 'MVLR' in fileName:
        list_MVLR.append(data_file)
    elif 'MVPR' in fileName:
        list_MVPR.append(data_file)
    elif 'FNO' in fileName:
        list_FNO.append(data_file)

## Create one DF per fitting type
all_MVLR = pd.concat(list_MVLR, axis = 0, ignore_index = True)
all_MVPR = pd.concat(list_MVPR, axis = 0, ignore_index = True)
all_FNO  = pd.concat(list_FNO, axis = 0, ignore_index = True)

## Save evaluation results for each model type
all_MVLR.to_csv(dir_MVLR)
all_MVPR.to_csv(dir_MVPR)
all_FNO.to_csv(dir_FNO)
```
